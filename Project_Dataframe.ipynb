```python
import pandas as pd

info_df = pd.read_excel("/content/hd2025.xlsx")
numbers_df = pd.read_excel("/content/ef_a_dist_IRE_Main.xlsx")
```

```python
df = pd.merge(info_df, numbers_df, on="UNITID", how="inner")

df
```

```python
def optimize_dtypes(df):
    for col in df.columns:
        # Convert text columns with repetitive values to 'category'
        if df[col].dtype == 'object':
            if df[col].nunique() / len(df) < 0.5:
                df[col] = df[col].astype('category')
        # Downcast integers
        elif df[col].dtype == 'int64':
            df[col] = pd.to_numeric(df[col], downcast='integer')
    return df

# Downsize the df for frame_to_hyper.
df = optimize_dtypes(df)
```

```python
# Force columns to be treated as text/string data

text_columns = ['ZIP', 'UEIS', 'OPEID']

for col in text_columns:
  df[col] = df[col].astype(str)

df['CLOSEDAT'] = pd.to_datetime(df['CLOSEDAT'], errors='coerce')
```
